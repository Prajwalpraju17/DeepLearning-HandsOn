```python
errors = []
predictions = []
```

```python
for i in range(epochs):
  y_pred = x1 * w1 + x2 * w2


  error = (y_actual - y_pred) ** 2

  errors.append(error)
  predictions.append(y_pred)


  d_error_d_ypred = -2 * (y_actual - y_pred)


  grad_w1 = d_error_d_ypred * x1
  grad_w2 = d_error_d_ypred * x2


  w1 = w1 - learning_rate * grad_w1
  w2 = w2 - learning_rate * grad_w2

  print(f"Epoch {i+1}: y_pred = {y_pred:.2f}, error = {error:.2f}")
```

Sigmoid Function (Definition)

The sigmoid function is a commonly used activation function in neural networks that converts any real-valued input into a value between 0 and 1.

Tanh Function (Definition)

The tanh (hyperbolic tangent) function is an activation function used in neural networks that maps input values to a range between -1 and 1.

ReLU Function (Definition)

The ReLU (Rectified Linear Unit) is a widely used activation function in neural networks that outputs the input directly if it is positive, otherwise it returns zero.
